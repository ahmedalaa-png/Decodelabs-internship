**Goal:** Clean a raw e-commerce dataset by handling missing values, duplicates, and incorrect data.

**Dataset:** E-Commerce Orders Dataset (1220 rows, 14 columns)


## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully ✅')
```

```python
df = pd.read_csv('ecommerce_raw_messy.csv')

print('Dataset Shape:', df.shape)
print('\nFirst 5 rows:')
df.head()
```

## Step 3: Initial Data Audit

```python
print('=== DATA TYPES ===')
print(df.dtypes)
print('\n=== MISSING VALUES ===')
print(df.isnull().sum())
print('\n=== DUPLICATE ROWS ===')
print('Total duplicates:', df.duplicated().sum())
```

## Phase 1: Handle Missing Values (Strategic Imputation)

```python
# CouponCode is missing for some orders - this means no coupon was used
# We fill with 'NO_COUPON' instead of deleting rows

missing_before = df['CouponCode'].isnull().sum()
df['CouponCode'] = df['CouponCode'].fillna('NO_COUPON')
missing_after = df['CouponCode'].isnull().sum()

print(f'Missing CouponCode before: {missing_before}')
print(f'Missing CouponCode after:  {missing_after}')
print('✅ Missing values handled - filled with NO_COUPON')
```

## Phase 2: Remove Duplicates (Integrity Audit)

```python
rows_before = len(df)
df = df.drop_duplicates()
rows_after = len(df)

print(f'Rows before removing duplicates: {rows_before}')
print(f'Rows after removing duplicates:  {rows_after}')
print(f'Duplicates removed: {rows_before - rows_after}')
print('✅ Duplicates removed successfully')
```

## Phase 3: Fix Data Formats (Speak One Language)

```python
# 3A: Fix Date column - remove INVALID_DATE rows
invalid_dates_before = (df['Date'] == 'INVALID_DATE').sum()
df = df[df['Date'] != 'INVALID_DATE']
df['Date'] = pd.to_datetime(df['Date'])

print(f'Invalid dates removed: {invalid_dates_before}')
print(f'Date dtype now: {df["Date"].dtype}')
print('✅ Date column fixed to ISO 8601 format (YYYY-MM-DD)')
```

```python
# 3B: Fix negative UnitPrice - remove invalid rows
negative_prices = (df['UnitPrice'] < 0).sum()
df = df[df['UnitPrice'] > 0]

print(f'Negative price rows removed: {negative_prices}')
print('✅ UnitPrice column cleaned - all values positive')
```

```python
# 3C: Fix Product column - strip whitespace and proper case
df['Product'] = df['Product'].str.strip().str.title()

print('Unique products after cleaning:')
print(df['Product'].unique())
print('✅ Product column cleaned - proper case and trimmed')
```

```python
# 3D: Round numeric columns to 2 decimal places
df['UnitPrice'] = df['UnitPrice'].round(2)
df['TotalPrice'] = df['TotalPrice'].round(2)

print('✅ Numeric precision fixed to 2 decimal places')
```

## Step 4: Verification Gate ✅
**Requirement: 0% Error Rate on Unique IDs and Date Formats**

```python
print('=== VERIFICATION GATE ===')
print(f'Duplicate OrderIDs: {df["OrderID"].duplicated().sum()} (must be 0)')
print(f'Invalid dates:      {df["Date"].isnull().sum()} (must be 0)')
print(f'Negative prices:    {(df["UnitPrice"] < 0).sum()} (must be 0)')
print(f'Missing values:     {df.isnull().sum().sum()} (must be 0)')
print(f'\nFinal dataset shape: {df.shape}')
print('\n✅ ALL CHECKS PASSED - Dataset is clean!')
```
